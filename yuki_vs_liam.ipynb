```python
import pandas as pd

def to_seconds(time_str):
    minutes, seconds = map(float, time_str.split(':'))
    return minutes * 60 + seconds

data = [
    {'Name': 'USA', 'Time_Liam': to_seconds('1:33.339'), 'Time_Yuki': to_seconds('1:33.795')},
    {'Name': 'MEXICO', 'Time_Liam': to_seconds('1:17.380'), 'Time_Yuki': to_seconds('1:17.182')},
    {'Name': 'BRASIL', 'Time_Liam': to_seconds('1:30.758'), 'Time_Yuki': to_seconds('1:29.172')},
    {'Name': 'LAS VEGAS', 'Time_Liam': to_seconds('1:34.087'), 'Time_Yuki': to_seconds('1:33.789')},
    {'Name': 'QATAR', 'Time_Liam': to_seconds('1:22.411'), 'Time_Yuki': to_seconds('1:22.364')},
    {'Name': 'ABU DHABI', 'Time_Liam': to_seconds('1:23.733'), 'Time_Yuki': to_seconds('1:23.735')},
]

df = pd.DataFrame(data)
df
```

```python
# Calculate the percentage difference between 'Time' and 'Time_Max'
df['Percentage_Difference'] = ((df['Time_Liam'] - df['Time_Yuki']) / (df['Time_Liam'] + df['Time_Yuki'])/2) * 100

# Group by 'Name' and calculate the average percentage difference
average_percentage_difference = df['Percentage_Difference'].mean()

# Group by 'Name' and calculate the standard deviation of the percentage difference
std_percentage_difference = df['Percentage_Difference'].std()

print(f"Average Percentage Difference: {average_percentage_difference:.2f}%")
print(f"Standard Deviation of Percentage Difference: {std_percentage_difference:.2f}%")
```

```python
import plotly.graph_objects as go

# Create a list of names in the desired order
names = df.Name.tolist()

# Create a list of colors
color = 'darkblue'

fig = go.Figure()
W = 1200
H = 600

for name in names:
    value = df.Percentage_Difference[df.Name == name].values[0]  # Extract single value
    label = "YUKI" if value > 0 else "LIAM"

    fig.add_trace(go.Bar(
        x=[name],
        y=[value],
        name=name,
        marker_color=color,
        text=label,  # Add label
        textposition="inside",  # Ensure text is inside the bars
        insidetextanchor="middle",
        textfont=dict(color="white")  # Make text white
    ))

fig.update_layout(
    title='How Laim and Yuki performed in Q1 in the same car',
    title_x=0.5,  # Center the title
    yaxis_title='Q1 Gap Liam - Yuki (%)',
    showlegend=False,
    plot_bgcolor='white',
    width=W,
    height=H
)

fig.update_yaxes(range=[-0.5, 0.5], dtick=0.25)
fig.update_yaxes(zeroline=True, zerolinecolor=color, zerolinewidth=.1, showline=True, linewidth=1, linecolor='black', ticks='outside')

fig.show()
```
